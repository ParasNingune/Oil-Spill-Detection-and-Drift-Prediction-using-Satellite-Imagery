```python
import rasterio
import numpy as np

# Read VV
with rasterio.open("/Users/parasningune/Downloads/zommed/2026-03-14-00:00_2026-03-14-23:59_Sentinel-1_IW_VV+VH_VV_(Raw).tiff") as vv_src:
    vv = vv_src.read(1)
    profile = vv_src.profile

# Read VH
with rasterio.open("/Users/parasningune/Downloads/zommed/2026-03-14-00:00_2026-03-14-23:59_Sentinel-1_IW_VV+VH_VH_(Raw).tiff") as vh_src:
    vh = vh_src.read(1)

# Stack bands
stacked = np.stack([vv, vh], axis=0)

# Update profile
profile.update(count=2)

# Save as multi-band TIFF
with rasterio.open("./combined.tif", "w", **profile) as dst:
    dst.write(stacked)

print("Combined TIFF created!")
```
